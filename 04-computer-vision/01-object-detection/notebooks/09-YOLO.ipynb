# Training Results

After training is completed, the results are stored in a **Result** object.

The most important evaluation metrics are:

---

## Precision

**Precision** shows:

> Among all objects detected by the model, how many predictions are actually correct?

Formula:

```

TP / (TP + FP)

```

High Precision means fewer **False Positive** detections.

---

## Recall

**Recall** shows:

> Among all real objects in the image, how many objects were successfully detected by the model?

Formula:

```

TP / (TP + FN)

```

High Recall means the model misses fewer objects.

---

## mAP@50

The mean Average Precision calculated at:

```

IoU = 0.5

```

---

## mAP@50-95

**mAP@50-95** evaluates the model at multiple IoU thresholds instead of using only IoU=0.5.

The model is evaluated at 10 different thresholds:

```

IoU = 0.50
IoU = 0.55
IoU = 0.60
IoU = 0.65
IoU = 0.70
IoU = 0.75
IoU = 0.80
IoU = 0.85
IoU = 0.90
IoU = 0.95

```

Then, the Average Precision for each threshold is calculated and averaged:

$$ mAP@50-95 =
(AP0.50 + AP0.55 + ... + AP0.95) / 10$$

---

## Why is mAP@50-95 More Strict?

Assume the model predicts a bounding box with:

```

IoU = 0.52

```

Evaluation:

- In **mAP@50**, because `0.52 > 0.50`, the prediction is considered **True Positive**.
- In **mAP@75**, because `0.52 < 0.75`, the prediction is considered incorrect.
- In **mAP@95**, it is also rejected.

As the IoU threshold increases, the model must place the bounding box more accurately on the object.

Example:

| IoU Threshold | Result |
|---------------|--------|
| 0.50          | ✅ Accepted |
| 0.55          | ❌ Rejected |
| 0.60          | ❌ Rejected |
| ...           | ❌ Rejected |

In this case, the model can achieve a high **mAP@50**, but its **mAP@50-95** will be lower.
```
