```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img1_bgr = cv2.imread('dataset/Segmentation/Glioma/enh_209.png')
```

```python
img1 = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2RGB)
```

```python
img1.shape
```

```python
plt.imshow(img1)
plt.axis('off')
plt.show()
```

```python
mask1 = cv2.imread('dataset/Segmentation/Glioma/enh_209_mask.png', cv2.COLOR_BGR2GRAY)
```

```python
mask1.shape
```

```python
plt.imshow(mask1, cmap='gray')
plt.axis('off')
plt.show()
```

```python
_, mask1 = cv2.threshold(mask1, 127, 255, cv2.THRESH_BINARY)
```

```python
plt.imshow(mask1, cmap='gray')
plt.axis('off')
plt.show()
```

```python
mask_bool = (mask1 == 255)

red_mask = np.zeros_like(img1)
red_mask[mask_bool] = [255,0,255]

image_with_highlight = img1.copy()

alpha = 0.2

image_with_highlight[mask_bool] = cv2.addWeighted(
    img1[mask_bool], 1 - alpha,
    red_mask[mask_bool], alpha, 0
)
```

```python
plt.imshow(image_with_highlight)
plt.axis('off') 
plt.show()
```

# Meningioma

```python
img2_bgr = cv2.imread('dataset/Segmentation/Meningioma/enh_1.png')
```

```python
img2 = cv2.cvtColor(img2_bgr, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img2)
plt.axis('off') 
plt.show()
```

```python
mask2 = cv2.imread('dataset/Segmentation/Meningioma/enh_1_mask.png', cv2.COLOR_BGR2GRAY)
```

```python
_, mask2 = cv2.threshold(mask2, 127, 255, cv2.THRESH_BINARY)
```

```python
plt.imshow(mask2, cmap='gray')
plt.axis('off')
plt.show()
```

```python
mask_bool = (mask2 == 255)

red_mask = np.zeros_like(img2)
red_mask[mask_bool] = [255,0,0]

image_with_highlight = img2.copy()

alpha = 0.5

image_with_highlight[mask_bool] = cv2.addWeighted(
    img2[mask_bool], 1 - alpha,
    red_mask[mask_bool], alpha, 0
)
```

```python
plt.imshow(image_with_highlight)
plt.axis('off')
plt.show()
```
